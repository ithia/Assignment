```python
def count_words():
    with open('document.txt', 'r') as file:
        content = file.read()
        words = content.split()
        return len(words)
print("Number of words:", count_words())
```

```python
def replace_word(old_word, new_word):
    with open('data.txt', 'r') as file:
        content = file.read()
    new_content = content.replace(old_word, new_word)
    with open('data.txt', 'w') as file:
        file.write(new_content)
replace_word('hello', 'hi')
print("Word replaced")
```

```python
def reverse_file():
    with open('reverse.txt', 'r') as file:
        lines = file.readlines()
    for i in range(len(lines) - 1, -1, -1):
        print(lines[i].strip())
reverse_file()
```

```python
def file_stats():
    with open('stats.txt', 'r') as file:
        content = file.read()
    lines = content.splitlines()
    words = content.split()
    total_lines = len(lines)
    total_words = len(words)
    total_chars = len(content)
    return total_lines, total_words, total_chars
l, w, c = file_stats()
print("Lines:", l)
print("Words:", w)
print("Characters:", c)
```

```python
import numpy as np
arr = np.random.randint(1, 50, size=(5, 5))
print("Array:")
print(arr)
top = arr[0, :]
bottom = arr[-1, :]
left = arr[1:-1, 0]
right = arr[1:-1, -1]
border = np.concatenate((top, right, bottom[::-1], left[::-1]))
print("Border elements:")
print(border)
```

```python
data = {
    'A': [12, np.nan, 25, 30, 18],
    'B': [20, 35, np.nan, 40, 25],
    'C': [15, 22, 28, np.nan, 35]
}

df = pd.DataFrame(data)

print("Before replacing:")
print(df)

df = df.fillna(df.mean())

print("\nAfter replacing:")
print(df)
```

```python
data = {
    'Product': ['Pen', 'Book', 'Pencil', 'Bag', 'Scale', 'Bottle'],
    'Category': ['Stationery', 'Education', 'Stationery', 'Accessories', 'Stationery', 'Accessories'],
    'Sales': [100, 250, 80, 500, 60, 300]
}

df = pd.DataFrame(data)

total = df.groupby('Category')['Sales'].sum()

print("Total Sales:")
print(total)
```

```python
df1 = pd.DataFrame({
    'ID': [101, 102, 103, 104],
    'Name': ['sia', 'shreya', 'vedh', 'Neha']
})

df2 = pd.DataFrame({
    'ID': [101, 102, 103, 104],
    'Marks': [85, 90, 78, 95]
})

merged = pd.merge(df1, df2, on='ID')

print("First DataFrame:")
print(df1)

print("\nSecond DataFrame:")
print(df2)

print("\nMerged DataFrame:")
print(merged)
```

```python
arrays = [
    ['east', 'east', 'north', 'north'],
    ['Apple', 'Mango', 'Apple', 'Mango']
]

index = pd.MultiIndex.from_arrays(arrays, names=['Region', 'Fruit'])

df = pd.DataFrame({
    'Quantity': [40, 25, 35, 50],
    'Price': [100, 80, 120, 90]
}, index=index)

print("MultiIndex DataFrame:")
print(df)

print("\nSelecting north:")
print(df.loc['north'])

print("\nSelecting Apple:")
print(df.xs('Apple', level='Fruit'))

print("\nSlicing Data:")
print(df.iloc[0:2])
```

```python
data = {
    'Category': ['food', 'food', 'food', 'clothes', 'clothes', 'clothes'],
    'SubCategory': ['Rice', 'Wheat', 'Rice', 'Shirt', 'Pants', 'Shirt'],
    'Amount': [100, 150, 200, 300, 250, 400]
}

df = pd.DataFrame(data)

result = df.groupby(['Category', 'SubCategory'])['Amount'].sum()

print("Sum of values:")
print(result)
```

```python
data = np.random.randint(5, 30, size=(6, 3))

df = pd.DataFrame(data, columns=['S', 'K', 'M'])

print("Original DataFrame:")
print(df)

df['Total'] = df.apply(lambda row: row['S'] + row['K'] + row['M'], axis=1)

print("\nDataFrame with Total:")
print(df)
```
